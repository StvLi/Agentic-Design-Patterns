# 第 15 章 · 两种交互方式：同步 vs 流式

这一格是**两份 JSON-RPC 消息体**，不是可执行代码。

| 书里的方法 | 语义 | 现在的名字（a2a-sdk 1.3.0） |
|---|---|---|
| `sendTask` | 发起任务，等最终结果 | `SendMessage` |
| `sendTaskSubscribe` | 发起任务，订阅过程事件 | `SendStreamingMessage` |

关键字段：

- `id` / `sessionId`：任务标识与会话标识（多轮上下文靠 `sessionId` 串起来）
- `message.parts[]`：内容分片，`type: text` 是文本
- `acceptedOutputModes`：客户端能接收的模态
- `historyLength`：要带多少历史（控制上下文长度）

另一处与书里不同的地方：请求必须带 `A2A-Version: 1.0` 头，否则服务端直接回
`A2A version '0.3' is not supported`。

实测两种模式的差别（同步 1 个事件 vs 流式 7 个事件）见 `../../ch15_a2a/README.md`。


```python
#Synchronous Request Example

{
  "jsonrpc": "2.0",
  "id": "1",
  "method": "sendTask",
  "params": {
    "id": "task-001",
    "sessionId": "session-001",
    "message": {
      "role": "user",
      "parts": [
        {
          "type": "text",
          "text": "What is the exchange rate from USD to EUR?"
        }
      ]
    },
    "acceptedOutputModes": ["text/plain"],
    "historyLength": 5
  }
}

# Streaming Request Example
 {
  "jsonrpc": "2.0",
  "id": "2",
  "method": "sendTaskSubscribe",
  "params": {
    "id": "task-002",
    "sessionId": "session-001",
    "message": {
      "role": "user",
      "parts": [
        {
          "type": "text",
          "text": "What's the exchange rate for JPY to GBP today?"
        }
      ]
    },
    "acceptedOutputModes": ["text/plain"],
    "historyLength": 5
  }
}
```
